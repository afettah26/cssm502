```python

## 1) Number Guessing Game
## Draw a random integer from 1 to 20 and let the user guess until correct
## answering "higher" or "lower" after each wrong guess

#Answer


import random # Rastgele sayı üretme fonksiyonlarını kullanmak için

def play_guessing_game():  #tahmin oyunu için ana fonksiyon
    target = random.randint(1, 20) # random kütüphanesinin içindeki randint fonksiyonunu çağırır.
                                   # bu fonskiyon belirttiğimiz aralıkta rastgele int üretir
    print("I've picked a number between 1 and 20. Try to guess it!")

    while True:                    #oyuncu doğru sayıyı tahmin edene kadar döngü oluşturur
        try:
            guess = int(input("Enter your guess: "))
        except ValueError:        #oyuncu int dışında değer girerse bu kısım hatayı yakalar
            print("Please enter a valid integer.")
            continue              # döngüyü devam ettiriri
        if guess < target:
            print("Higher!")
        elif guess > target:
            print("Lower!")
        else:
            print("Correct! You got it!")
            break

play_guessing_game()
```

```python
## 2) Search insert position.
## Given a sorted array of distinct integers and a target, return the target's
## index, or the index where it would be inserted



## Answer:

def search_insert(nums, target):      # Listenin her bir elemanını sırayla kontrol etmek için fonksiyon
    for i in range(len(nums)):
        # Eğer sıradaki sayı hedeften büyük veya eşitse,
        # hedef sayının bulunması veya eklenmesi gereken yer tam olarak bu indekstir (i).
        if nums[i] >= target:
            return i

    # Eğer döngü biter ve aradığımızı bulamazsak,
    # demek ki hedef sayı listedeki tüm sayılardan büyüktür.
    # Bu yüzden listenin en sonuna (listenin uzunluğu kadar olan indekse) eklenmelidir.
    return len(nums)

# Örnek Testler:
numbers = [1, 3, 5, 6]

print("For 5 the index:", search_insert(numbers, 5))   # Output: 2 (Listedeki yeri)
print("For 2 the index:", search_insert(numbers, 2))   # Output: 1 (1 ile 3 arasına eklenmeli)
print("For 7 the index:", search_insert(numbers, 7))   # Output: 4 (En sona eklenmeli)
```

```python
## 3) Add binary strings
## Return the sum of two binary strings as a binary string ("11" + "1" -> "100")


## Answer:

def add_binary(a, b):
    # 1. Adım: İkilik tabandaki stringleri normal tam sayılara çevirme
    num1 = int(a, 2) # , 2 -> int 2lik sistemden normal sayıya çevirme
    num2 = int(b, 2)

    # 2. Adım: Sayıları toplama
    total_sum = num1 + num2

    # 3. Adım: Toplamı tekrar ikilik tabana çevirme ve başındaki '0b' kısmını temizle
    binary_sum = bin(total_sum)[2:] #bin fonksiyonu normal int 2lik sistemine çevirir

    return binary_sum

#Examples:
print("11 + 1 =", add_binary("11", "1"))
print("1010 + 1011 =", add_binary("1010", "1011"))
```

```python
## 4)Single number
## In an array where every element appears twice except one, find that one


## Answer:

def single_number(sayi):     # Listedeki her bir sayıyı sırayla kontrol ediyoruz
    for x in sayi:
        if sayi.count(x) == 1:   #Eğer sayı bir kere geçiyorsa
            return x

# Examples:
sayilar1 = [2, 2, 1]              # 2'ler çift, 1 yalnız
sayilar2 = [4, 1, 2, 1, 2]        # 1 ve 2 çift, 4 yalnız

print(sayilar1, "içinde bir kere geçen sayı:", single_number(sayilar1))
print(sayilar2, "içinde bir kere geçen sayı:", single_number(sayilar2))
```

```python
## 5) Repeated DNA Sequences
## Return all 10-letter substrings that occur more than once in DNA string
## ("AAAAACCCCCAAAAACCCCCAAAAAGGGTTT" -> ["AAAAACCCCC", "CCCCCAAAAA"])


## Answer:

def find_repeated_dna_seq(s):
    seen = set()       # Daha önce gördüğümüz 10'lu dizilimleri kaydetmek için
    repeat = set()   # Birden fazla kez gördüğümüz dizilimleri kaydetmek için

    # DNA diziliminin üzerinde 10 harf olacak şekilde ilerliyor
    for i in range(len(s) - 9):
        low_seq = s[i:i+10]  # i. indeksten başlayıp 10 karakter alıyoruz

        # Eğer bu 10'lu dizilimi daha önce gördüysek, tekrar ediyordur
        if low_seq in seen:
            repeat.add(low_seq)
        else:
            # İlk defa gördüysek görülenler listemize ekliyoruz
            seen.add(low_seq)

    return list(repeat)

# example
dna = "AAAAACCCCCAAAAACCCCCAAAAAGGGTTT"
result = find_repeated_dna_seq(dna)

print("Tekrar eden 10 harfli dizilimler:", result)
```
